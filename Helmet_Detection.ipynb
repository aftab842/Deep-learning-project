```python
import os
import random
import shutil

random.seed(42)

train_images = "/content/train/images"
train_labels = "/content/train/labels"

valid_images = "/content/valid/images"
valid_labels = "/content/valid/labels"

os.makedirs(valid_images, exist_ok=True)
os.makedirs(valid_labels, exist_ok=True)

images = [
    f for f in os.listdir(train_images)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

random.shuffle(images)

# Move 2 images into validation
for img in images[:2]:
    shutil.move(
        os.path.join(train_images, img),
        os.path.join(valid_images, img)
    )

    label = os.path.splitext(img)[0] + ".txt"
    label_path = os.path.join(train_labels, label)

    if os.path.exists(label_path):
        shutil.move(
            label_path,
            os.path.join(valid_labels, label)
        )

print("Validation dataset created successfully!")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model.train(
    data="/content/data.yaml",
    epochs=100,
    imgsz=640,
    batch=4
)
```
